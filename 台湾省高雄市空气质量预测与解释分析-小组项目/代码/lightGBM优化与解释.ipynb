```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr
import lightgbm as lgb
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler, RobustScaler
from sklearn.multioutput import MultiOutputRegressor
import matplotlib.pyplot as plt
import shap
import warnings
from typing import Dict, List, Tuple, Optional
from dataclasses import dataclass, field
import logging
import joblib
import optuna
import time

warnings.filterwarnings('ignore')

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s'
)
logger = logging.getLogger(__name__)
optuna.logging.set_verbosity(optuna.logging.WARNING)


def check_gpu_availability():
    """检测GPU是否可用"""
    try:
        test_params = {'device': 'gpu', 'gpu_platform_id': 0, 'gpu_device_id': 0}
        test_data = lgb.Dataset(np.random.rand(100, 10), label=np.random.rand(100))
        lgb.train(test_params, test_data, num_boost_round=1, verbose_eval=False)
        print("✅ GPU 可用")
        return True
    except:
        print("⚠️ GPU 不可用，使用 CPU")
        return False

USE_GPU = check_gpu_availability()


@dataclass
class ModelConfig:
    """模型配置"""
    lookback_hours: int = 48
    forecast_hours: int = 24
    max_lag: int = 24
    
    test_size: float = 0.15
    val_size: float = 0.15
    
    lgb_params: dict = field(default_factory=dict)
    num_boost_round: int = 500
    early_stopping_rounds: int = 50
    use_gpu: bool = USE_GPU
    
    use_optuna: bool = True
    n_trials: int = 50
    optuna_timeout: int = 3600
    optuna_sample_ratio: float = 0.5
    optuna_eval_hours: List[int] = field(default_factory=lambda: [0, 5, 11, 23])
    
    # SHAP配置
    use_shap: bool = True
    shap_sample_size: int = 500
    shap_waterfall_hour: int = 0  # Waterfall图使用的小时（第1小时）
    
    def __post_init__(self):
        if not self.lgb_params:
            self.lgb_params = self._get_default_params()
    
    def _get_default_params(self) -> dict:
        params = {
            'objective': 'regression',
            'metric': 'rmse',
            'boosting_type': 'gbdt',
            'verbose': -1,
            'random_state': 42,
            'n_jobs': -1,
            'num_leaves': 31,
            'max_depth': 6,
            'learning_rate': 0.05,
        }
        if self.use_gpu:
            params.update({'device': 'gpu', 'gpu_platform_id': 0, 'gpu_device_id': 0})
        return params
    
    def print_summary(self):
        train_pct = (1 - self.test_size - self.val_size) * 100
        print(f"\n{'='*60}")
        print("📋 Model Configuration")
        print(f"{'='*60}")
        print(f"   Data Split: Train {train_pct:.0f}% | Val {self.val_size*100:.0f}% | Test {self.test_size*100:.0f}%")
        print(f"   Lookback: {self.lookback_hours}h")
        print(f"   Forecast: {self.forecast_hours}h")
        if self.use_optuna:
            print(f"   Optuna: {self.n_trials} trials, {self.optuna_sample_ratio*100:.0f}% subsample")
        if self.use_shap:
            print(f"   SHAP: {self.shap_sample_size} samples, Average over all {self.forecast_hours} hours")
        print(f"{'='*60}")


class DataLoader:
    @staticmethod
    def load_data(data_path: str) -> pd.DataFrame:
        data = pd.read_csv(data_path)
        data['date'] = pd.to_datetime(data['date'])
        data = data.sort_values(['sitename', 'date']).reset_index(drop=True)
        return data


class CorrelationAnalyzer:
    @staticmethod
    def analyze_lag_correlation(target_series, neighbor_series, max_lag):
        correlations = []
        for lag in range(max_lag + 1):
            shifted = neighbor_series.shift(lag) if lag > 0 else neighbor_series
            valid_idx = target_series.notna() & shifted.notna()
            if valid_idx.sum() < 10:
                correlations.append(0)
            else:
                try:
                    corr, _ = pearsonr(target_series[valid_idx], shifted[valid_idx])
                    correlations.append(corr if not np.isnan(corr) else 0)
                except:
                    correlations.append(0)
        best_lag = np.argmax(correlations)
        return best_lag, correlations[best_lag], correlations[0]


class EnhancedFeatureEngine:
    @staticmethod
    def create_comprehensive_features(df_base: pd.DataFrame, neighbor_sites: List[str]) -> pd.DataFrame:
        df = df_base.copy()
        
        df['hour'] = df.index.hour
        df['month'] = df.index.month
        df['dayofweek'] = df.index.dayofweek
        
        df['hour_sin'] = np.sin(2 * np.pi * df['hour'] / 24)
        df['hour_cos'] = np.cos(2 * np.pi * df['hour'] / 24)
        df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12)
        df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12)
        df['day_sin'] = np.sin(2 * np.pi * df['dayofweek'] / 7)
        df['day_cos'] = np.cos(2 * np.pi * df['dayofweek'] / 7)
        df['is_weekend'] = (df['dayofweek'] >= 5).astype(int)
        
        if 'winddirec' in df.columns and 'windspeed' in df.columns:
            df['wind_rad'] = np.deg2rad(df['winddirec'])
            df['Wind_X'] = df['windspeed'] * np.cos(df['wind_rad'])
            df['Wind_Y'] = df['windspeed'] * np.sin(df['wind_rad'])
        
        for neighbor in neighbor_sites:
            wind_dir_col = f'{neighbor}_winddirec'
            wind_speed_col = f'{neighbor}_windspeed'
            if wind_dir_col in df.columns and wind_speed_col in df.columns:
                df[f'{neighbor}_wind_rad'] = np.deg2rad(df[wind_dir_col])
                df[f'{neighbor}_Wind_X'] = df[wind_speed_col] * np.cos(df[f'{neighbor}_wind_rad'])
                df[f'{neighbor}_Wind_Y'] = df[wind_speed_col] * np.sin(df[f'{neighbor}_wind_rad'])
        
        for lag in [1, 3, 6, 12, 24]:
            if 'AQI' in df.columns:
                df[f'AQI_lag_{lag}'] = df['AQI'].shift(lag)
            if 'PM2.5' in df.columns:
                df[f'PM25_lag_{lag}'] = df['PM2.5'].shift(lag)
        
        for window in [6, 12, 24]:
            if 'AQI' in df.columns:
                df[f'AQI_roll_mean_{window}'] = df['AQI'].rolling(window).mean()
                df[f'AQI_roll_std_{window}'] = df['AQI'].rolling(window).std()
        
        neighbor_aqi_cols = [f'{n}_AQI' for n in neighbor_sites if f'{n}_AQI' in df.columns]
        if neighbor_aqi_cols:
            df['neighbor_AQI_mean'] = df[neighbor_aqi_cols].mean(axis=1)
            df['neighbor_AQI_std'] = df[neighbor_aqi_cols].std(axis=1)
        
        return df
    
    @staticmethod
    def select_final_features(df: pd.DataFrame, neighbor_sites: List[str]) -> List[str]:
        features_to_keep = [
            'PM2.5', 'PM10', 'O3', 'SO2', 'NOx', 'CO', 'AQI',
            'Wind_X', 'Wind_Y',
            'hour_sin', 'hour_cos', 'month_sin', 'month_cos', 
            'day_sin', 'day_cos', 'is_weekend',
            'AQI_lag_1', 'AQI_lag_3', 'AQI_lag_6', 'AQI_lag_12', 'AQI_lag_24',
            'PM25_lag_1', 'PM25_lag_3', 'PM25_lag_6', 'PM25_lag_12', 'PM25_lag_24',
            'AQI_roll_mean_6', 'AQI_roll_mean_12', 'AQI_roll_mean_24',
            'AQI_roll_std_6', 'AQI_roll_std_12', 'AQI_roll_std_24',
            'neighbor_AQI_mean', 'neighbor_AQI_std'
        ]
        
        for neighbor in neighbor_sites:
            features_to_keep.extend([
                f'{neighbor}_PM2.5',
                f'{neighbor}_AQI',
                f'{neighbor}_Wind_X',
                f'{neighbor}_Wind_Y'
            ])
        
        final_cols = [c for c in features_to_keep if c in df.columns]
        return final_cols


class FastOptunaOptimizer:
    def __init__(self, X_train: np.ndarray, y_train: np.ndarray, 
                 X_val: np.ndarray, y_val: np.ndarray, config: ModelConfig):
        self.config = config
        self.best_params = None
        self.study = None
        
        sample_size = int(len(X_train) * config.optuna_sample_ratio)
        np.random.seed(42)
        indices = np.random.choice(len(X_train), sample_size, replace=False)
        
        self.X_train = X_train[indices]
        self.y_train = y_train[indices]
        self.X_val = X_val
        self.y_val = y_val
        
        self.eval_hours = config.optuna_eval_hours 
        
        print(f"\n📊 Optuna Optimizer Initialized:")
        print(f"   Training samples: {len(self.X_train)} ({config.optuna_sample_ratio*100:.0f}% of {len(X_train)})")
        print(f"   Validation samples: {len(self.X_val)}")
        print(f"   Evaluation hours: {[h+1 for h in self.eval_hours]}")
    
    def objective(self, trial: optuna.Trial) -> float:
        params = {
            'objective': 'regression',
            'metric': 'rmse',
            'boosting_type': 'gbdt',
            'verbose': -1,
            'random_state': 42,
            'n_jobs': -1,
            'num_leaves': trial.suggest_int('num_leaves', 20, 80),
            'max_depth': trial.suggest_int('max_depth', 4, 8),
            'min_child_samples': trial.suggest_int('min_child_samples', 20, 100),
            'lambda_l1': trial.suggest_float('lambda_l1', 0.001, 3.0, log=True),
            'lambda_l2': trial.suggest_float('lambda_l2', 0.001, 3.0, log=True),
            'feature_fraction': trial.suggest_float('feature_fraction', 0.5, 0.9),
            'bagging_fraction': trial.suggest_float('bagging_fraction', 0.5, 0.9),
            'bagging_freq': trial.suggest_int('bagging_freq', 3, 7),
            'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.08, log=True),
        }
        
        if self.config.use_gpu:
            params.update({'device': 'gpu', 'gpu_platform_id': 0, 'gpu_device_id': 0})
        
        n_estimators = trial.suggest_int('n_estimators', 400, 1500)
        
        try:
            scores = []
            for h in self.eval_hours:
                train_data = lgb.Dataset(self.X_train, label=self.y_train[:, h])
                val_data = lgb.Dataset(self.X_val, label=self.y_val[:, h], reference=train_data)
                
                model = lgb.train(
                    params, train_data, num_boost_round=n_estimators,
                    valid_sets=[val_data], valid_names=['valid'],
                    callbacks=[
                        lgb.early_stopping(stopping_rounds=50, verbose=False),
                        lgb.log_evaluation(period=0)
                    ]
                )
                
                y_pred = model.predict(self.X_val)
                r2 = r2_score(self.y_val[:, h], y_pred)
                scores.append(r2)
                
                trial.report(np.mean(scores), h)
                if trial.should_prune():
                    raise optuna.TrialPruned()
            
            weights = [1.5, 1.2, 0.8, 0.5]
            weighted_r2 = np.average(scores, weights=weights)
            return weighted_r2
            
        except optuna.TrialPruned:
            raise
        except Exception as e:
            return float('-inf')
    
    def optimize(self) -> Dict:
        print("\n" + "=" * 70)
        print("🚀 Starting Optuna Hyperparameter Optimization")
        print("=" * 70)
        
        self.study = optuna.create_study(
            direction='maximize',
            sampler=optuna.samplers.TPESampler(seed=42, n_startup_trials=15),
            pruner=optuna.pruners.MedianPruner(n_warmup_steps=2)
        )
        
        initial_params = {
            'num_leaves': 31, 'max_depth': 6, 'min_child_samples': 50,
            'lambda_l1': 0.5, 'lambda_l2': 0.3,
            'feature_fraction': 0.7, 'bagging_fraction': 0.7, 'bagging_freq': 5,
            'learning_rate': 0.03, 'n_estimators': 800,
        }
        self.study.enqueue_trial(initial_params)
        
        start_time = time.time()
        
        def callback(study, trial):
            elapsed = time.time() - start_time
            if trial.number % 5 == 0 or trial.number < 5:
                r2_str = f"{trial.value:.4f}" if trial.value is not None else "N/A"
                best_str = f"{study.best_value:.4f}" if len(study.trials) > 0 and study.best_value else "N/A"
                print(f"   Trial {trial.number:3d} | R²={r2_str:>7} | Best={best_str} | Time={elapsed:.0f}s")
        
        self.study.optimize(
            self.objective, n_trials=self.config.n_trials,
            timeout=self.config.optuna_timeout,
            callbacks=[callback], show_progress_bar=True, n_jobs=1
        )
        
        self.best_params = self.study.best_params
        
        print("\n" + "=" * 70)
        print("✅ Optuna Optimization Completed!")
        print("=" * 70)
        print(f"   Best R²: {self.study.best_value:.4f}")
        print("\n📊 Best Parameters:")
        for key, value in self.best_params.items():
            print(f"   {key}: {value}")
        print("=" * 70)
        
        return self.best_params
    
    def get_best_lgb_params(self) -> Tuple[dict, int]:
        if self.best_params is None:
            raise ValueError("Please run optimize() first")
        
        params_copy = self.best_params.copy()
        n_estimators = params_copy.pop('n_estimators', 800)
        
        lgb_params = {
            'objective': 'regression', 'metric': 'rmse', 'boosting_type': 'gbdt',
            'verbose': -1, 'random_state': 42, 'n_jobs': -1,
            **params_copy
        }
        
        if self.config.use_gpu:
            lgb_params.update({'device': 'gpu', 'gpu_platform_id': 0, 'gpu_device_id': 0})
        
        return lgb_params, n_estimators


class FastPredictor:
    def __init__(self, data_path: str, target_site: str, config: Optional[ModelConfig] = None):
        self.data = DataLoader.load_data(data_path)
        self.all_sites = sorted(self.data['sitename'].unique())
        self.target_site = target_site
        self.config = config or ModelConfig()
        
        self.scaler_X = StandardScaler()
        self.scaler_y = RobustScaler()
        self.model = None
        self.optimizer = None
        self.best_params = None
        
        # 用于SHAP分析
        self.X_test_df = None
        self.feature_names = None
        
        logger.info(f"Predictor initialized: {target_site}")
        self.config.print_summary()
    
    def analyze_correlations(self) -> Dict:
        neighbor_sites = [s for s in self.all_sites if s != self.target_site]
        pivot_data = self.data.pivot(index='date', columns='sitename', values='AQI').sort_index()
        
        lag_config = {}
        print(f"\n{'='*60}")
        print(f"📊 Correlation Analysis: {self.target_site}")
        print(f"{'='*60}")
        
        for neighbor in neighbor_sites:
            if neighbor in pivot_data.columns:
                best_lag, best_corr, _ = CorrelationAnalyzer.analyze_lag_correlation(
                    pivot_data[self.target_site], pivot_data[neighbor], self.config.max_lag
                )
                lag_config[neighbor] = best_lag
                print(f"  {neighbor}: lag={best_lag}h, corr={best_corr:.4f}")
        
        print(f"{'='*60}")
        return lag_config
    
    def prepare_features(self, lag_config: Dict) -> pd.DataFrame:
        neighbor_sites = [s for s in self.all_sites if s != self.target_site]
        
        df_target = self.data[self.data['sitename'] == self.target_site].copy()
        df_target = df_target.set_index('date')
        
        df_neighbors = self.data[self.data['sitename'].isin(neighbor_sites)].copy()
        df_base = df_target.copy()
        
        for neighbor in neighbor_sites:
            df_neighbor = df_neighbors[df_neighbors['sitename'] == neighbor].copy()
            df_neighbor = df_neighbor.set_index('date')
            
            lag = lag_config.get(neighbor, 0)
            if lag > 0:
                df_neighbor = df_neighbor.shift(lag)
            
            neighbor_features = ['PM2.5', 'AQI', 'winddirec', 'windspeed']
            available_features = [f for f in neighbor_features if f in df_neighbor.columns]
            
            if available_features:
                df_neighbor_selected = df_neighbor[available_features].copy()
                df_neighbor_selected.columns = [f'{neighbor}_{col}' for col in available_features]
                df_base = df_base.join(df_neighbor_selected, how='left')
        
        df_base = df_base.ffill().bfill()
        df_base = EnhancedFeatureEngine.create_comprehensive_features(df_base, neighbor_sites)
        final_cols = EnhancedFeatureEngine.select_final_features(df_base, neighbor_sites)
        
        df_processed = df_base[final_cols].copy()
        df_processed = df_processed.dropna()
        
        logger.info(f"Feature engineering completed: {len(final_cols)} features, {len(df_processed)} samples")
        
        return df_processed
    
    def create_sequences(self, df_features: pd.DataFrame) -> Tuple[np.ndarray, np.ndarray]:
        df_features = df_features.sort_index()
        
        if 'AQI' not in df_features.columns:
            raise ValueError("Missing AQI column in data")
        
        # 保存特征名（用于SHAP）
        self.feature_names = df_features.columns.tolist()
        
        aqi_col_idx = df_features.columns.get_loc('AQI')
        data_array = df_features.values
        
        X_list, y_list = [], []
        
        input_len = self.config.lookback_hours
        output_len = self.config.forecast_hours
        total_len = len(data_array)
        
        for i in range(total_len - input_len - output_len + 1):
            X_window = data_array[i:i + input_len]
            y_window = data_array[i + input_len:i + input_len + output_len, aqi_col_idx]
            
            X_list.append(X_window)
            y_list.append(y_window)
        
        X = np.array(X_list)
        y = np.array(y_list)
        
        logger.info(f"Sequence creation completed: X shape = {X.shape}, y shape = {y.shape}")
        
        return X, y
    
    def prepare_data(self, X: np.ndarray, y: np.ndarray):
        total_samples = len(X)
        
        train_size = int(total_samples * 0.7)
        val_size = int(total_samples * 0.15)
        
        X_train_raw = X[:train_size]
        y_train_raw = y[:train_size]
        X_val_raw = X[train_size:train_size + val_size]
        y_val_raw = y[train_size:train_size + val_size]
        X_test_raw = X[train_size + val_size:]
        y_test_raw = y[train_size + val_size:]
        
        n_samples_train, n_timesteps, n_features = X_train_raw.shape
        
        self.scaler_X = StandardScaler()
        self.scaler_X.fit(X_train_raw.reshape(-1, n_features))
        
        X_train_scaled = self.scaler_X.transform(
            X_train_raw.reshape(-1, n_features)
        ).reshape(X_train_raw.shape)
        
        X_val_scaled = self.scaler_X.transform(
            X_val_raw.reshape(-1, n_features)
        ).reshape(X_val_raw.shape)
        
        X_test_scaled = self.scaler_X.transform(
            X_test_raw.reshape(-1, n_features)
        ).reshape(X_test_raw.shape)
        
        X_train_flat = X_train_scaled.reshape(len(X_train_scaled), -1)
        X_val_flat = X_val_scaled.reshape(len(X_val_scaled), -1)
        X_test_flat = X_test_scaled.reshape(len(X_test_scaled), -1)
        
        # 创建DataFrame用于SHAP（保留特征名）
        flat_feature_names = []
        for t in range(n_timesteps):
            for f in self.feature_names:
                flat_feature_names.append(f"{f}_t-{n_timesteps - t}")
        
        self.X_test_df = pd.DataFrame(X_test_flat, columns=flat_feature_names)
        
        self.scaler_y.fit(y_train_raw)
        y_train_scaled = self.scaler_y.transform(y_train_raw)
        y_val_scaled = self.scaler_y.transform(y_val_raw)
        y_test_scaled = self.scaler_y.transform(y_test_raw)
        
        return (
            X_train_flat, y_train_scaled,
            X_val_flat, y_val_scaled,
            X_test_flat, y_test_scaled,
            y_train_raw, y_val_raw, y_test_raw
        )
    
    def train_final_model(self, X_train, y_train, X_val, y_val, lgb_params, n_estimators):
        print(f"\n{'='*60}")
        print("🚀 Training Final Model (Train + Validation)")
        print(f"{'='*60}")
        
        X_full = np.vstack([X_train, X_val])
        y_full = np.vstack([y_train, y_val])
        
        print(f"   Total training samples: {len(X_full)}")
        print(f"   n_estimators: {n_estimators}")
        
        base_model = lgb.LGBMRegressor(**lgb_params, n_estimators=n_estimators)
        self.model = MultiOutputRegressor(base_model, n_jobs=-1)
        
        start = time.time()
        self.model.fit(X_full, y_full)
        print(f"✅ Training completed! Time: {time.time()-start:.1f}s")
    
    def predict(self, X: np.ndarray) -> np.ndarray:
        y_pred_scaled = self.model.predict(X)
        return self.scaler_y.inverse_transform(y_pred_scaled)
    
    def evaluate(self, y_true: np.ndarray, y_pred: np.ndarray) -> Dict:
        forecast_hours = self.config.forecast_hours
        
        hourly_metrics = []
        for h in range(forecast_hours):
            hourly_metrics.append({
                'hour': h + 1,
                'rmse': np.sqrt(mean_squared_error(y_true[:, h], y_pred[:, h])),
                'mae': mean_absolute_error(y_true[:, h], y_pred[:, h]),
                'r2': r2_score(y_true[:, h], y_pred[:, h])
            })
        
        hourly_df = pd.DataFrame(hourly_metrics)
        
        weights = np.linspace(1.5, 0.5, forecast_hours)
        weights /= weights.sum()
        weighted_r2 = np.average(hourly_df['r2'].values, weights=weights)
        
        return {
            'hourly': hourly_df,
            'overall': {
                'rmse': np.sqrt(mean_squared_error(y_true.flatten(), y_pred.flatten())),
                'mae': mean_absolute_error(y_true.flatten(), y_pred.flatten()),
                'r2_weighted': weighted_r2,
                'r2_flat': r2_score(y_true.flatten(), y_pred.flatten())
            },
            'period': {
                '短期(1-6h)': {'r2': r2_score(y_true[:,:6].flatten(), y_pred[:,:6].flatten())},
                '中期(7-12h)': {'r2': r2_score(y_true[:,6:12].flatten(), y_pred[:,6:12].flatten())},
                '长期(13-24h)': {'r2': r2_score(y_true[:,12:].flatten(), y_pred[:,12:].flatten())}
            }
        }
    
    def detailed_horizon_analysis(self, y_true: np.ndarray, y_pred: np.ndarray):
        """详细的分时段分析"""
        print("\n" + "=" * 70)
        print("📈 Hourly RMSE Analysis")
        print("=" * 70)
        
        step_rmse = np.sqrt(np.mean((y_true - y_pred) ** 2, axis=0))
        step_mae = np.mean(np.abs(y_true - y_pred), axis=0)
        
        print("\nHour   RMSE    MAE")
        print("-" * 30)
        for h in [1, 3, 6, 12, 18, 24]:
            print(f"{h:3d}h  {step_rmse[h-1]:6.2f}  {step_mae[h-1]:6.2f}")
        
        horizons = {
            'Ultra-Short (1-6h)': (0, 6),
            'Short (7-12h)': (6, 12),
            'Medium (13-18h)': (12, 18),
            'Long (19-24h)': (18, 24)
        }
        
        print("\n" + "=" * 70)
        print("📋 Detailed Performance Report by Period")
        print("=" * 70)
        
        horizon_results = {}
        for name, (start, end) in horizons.items():
            y_true_horizon = y_true[:, start:end]
            y_pred_horizon = y_pred[:, start:end]
            
            rmse_horizon = np.sqrt(mean_squared_error(y_true_horizon, y_pred_horizon))
            mae_horizon = mean_absolute_error(y_true_horizon, y_pred_horizon)
            r2_horizon = r2_score(y_true_horizon.flatten(), y_pred_horizon.flatten())
            
            horizon_results[name] = {
                'rmse': rmse_horizon,
                'mae': mae_horizon,
                'r2': r2_horizon
            }
            
            print(f"\n{name}:")
            print(f"   RMSE: {rmse_horizon:.4f}")
            print(f"   MAE : {mae_horizon:.4f}")
            print(f"   R²  : {r2_horizon:.4f}")
        
        print("=" * 70)
        
        return step_rmse, step_mae, horizon_results
    
    def plot_detailed_results(self, y_true: np.ndarray, y_pred: np.ndarray, 
                             step_rmse: np.ndarray, step_mae: np.ndarray):
        """绘制预测结果"""
        hours = np.arange(1, 25)
        
        fig = plt.figure(figsize=(18, 10))
        gs = fig.add_gridspec(2, 3, hspace=0.3, wspace=0.3)
        
        np.random.seed(42)
        sample_ids = np.random.choice(len(y_true), size=3, replace=False)
        
        for i, sample_id in enumerate(sample_ids):
            ax = fig.add_subplot(gs[0, i])
            ax.plot(hours, y_true[sample_id], label='Actual', 
                   marker='o', markersize=4, linewidth=2, color='blue')
            ax.plot(hours, y_pred[sample_id], label='Predicted', 
                   marker='x', markersize=4, linestyle='--', linewidth=2, color='red')
            ax.set_xlabel('Hours Ahead', fontsize=11)
            ax.set_ylabel('AQI', fontsize=11)
            ax.set_title(f'Sample {sample_id}', fontweight='bold', fontsize=12)
            ax.legend(loc='best')
            ax.grid(True, alpha=0.3)
        
        sample_ids_2 = np.random.choice(len(y_true), size=3, replace=False)
        
        for i, sample_id in enumerate(sample_ids_2):
            ax = fig.add_subplot(gs[1, i])
            ax.plot(hours, y_true[sample_id], label='Actual', 
                   marker='o', markersize=4, linewidth=2, color='blue')
            ax.plot(hours, y_pred[sample_id], label='Predicted', 
                   marker='x', markersize=4, linestyle='--', linewidth=2, color='red')
            ax.set_xlabel('Hours Ahead', fontsize=11)
            ax.set_ylabel('AQI', fontsize=11)
            ax.set_title(f'Sample {sample_id}', fontweight='bold', fontsize=12)
            ax.legend(loc='best')
            ax.grid(True, alpha=0.3)
        
        plt.suptitle(f'{self.target_site} - 24h AQI Forecast Results', 
                    fontsize=16, fontweight='bold', y=0.995)
        plt.savefig(f'{self.target_site}_24h_forecast_results.png', dpi=300, bbox_inches='tight')
        logger.info(f"Results plot saved: {self.target_site}_24h_forecast_results.png")
        plt.show()
    
    # ==================== SHAP Analysis Module (Average over all hours) ====================
    
    def analyze_shap(self, y_test_raw: np.ndarray, y_pred_test: np.ndarray):
        """
        ==================== SHAP Analysis (Average over All 24 Hours) ====================
        Calculate SHAP values for all hours and average them
        """
        if not self.config.use_shap:
            print("\n⚠️ SHAP analysis disabled")
            return None
        
        print("\n" + "=" * 100)
        print("【SHAP Value Calculation - Averaged Across All 24 Hours】")
        print("=" * 100)
        
        # Sample data
        sample_size = min(self.config.shap_sample_size, len(self.X_test_df))
        np.random.seed(42)
        sample_indices = np.random.choice(len(self.X_test_df), sample_size, replace=False)
        X_test_sample = self.X_test_df.iloc[sample_indices].reset_index(drop=True)
        y_test_sample = y_test_raw[sample_indices]
        y_pred_sample = y_pred_test[sample_indices]
        
        print(f"\n➤ SHAP Sample Size: {sample_size}")
        print(f"   Computing SHAP for all {self.config.forecast_hours} hours...")
        
        # ========== Calculate SHAP values for all 24 hours ==========
        all_shap_values = []
        all_expected_values = []
        
        for hour_idx in range(self.config.forecast_hours):
            hour_label = hour_idx + 1
            
            if hour_idx % 6 == 0 or hour_idx == self.config.forecast_hours - 1:
                print(f"   Computing Hour {hour_label}...")
            
            estimator = self.model.estimators_[hour_idx]
            
            try:
                explainer = shap.TreeExplainer(estimator)
                shap_values = explainer.shap_values(X_test_sample)
                expected_value = explainer.expected_value
            except Exception as e:
                print(f"   ⚠ TreeExplainer failed for Hour {hour_label}, using KernelExplainer...")
                background = shap.sample(X_test_sample, min(100, len(X_test_sample)), random_state=42)
                explainer = shap.KernelExplainer(estimator.predict, background)
                shap_values = explainer.shap_values(X_test_sample, nsamples=100)
                expected_value = explainer.expected_value
            
            all_shap_values.append(shap_values)
            all_expected_values.append(expected_value)
        
        print(f"✓ SHAP values computed for all {self.config.forecast_hours} hours")
        
        # ========== Average SHAP values across all hours ==========
        # Shape: (24, n_samples, n_features) -> (n_samples, n_features)
        averaged_shap_values = np.mean(all_shap_values, axis=0)
        averaged_expected_value = np.mean(all_expected_values)
        
        print(f"\n✓ Averaged SHAP values shape: {averaged_shap_values.shape}")
        print(f"  Averaged expected value: {averaged_expected_value:.4f}")
        
        # Calculate feature importance
        mean_abs_shap = np.abs(averaged_shap_values).mean(axis=0)
        feature_importance_df = pd.DataFrame({
            'Feature': X_test_sample.columns,
            'Mean_Abs_SHAP': mean_abs_shap
        }).sort_values('Mean_Abs_SHAP', ascending=False)
        
        print(f"\nTop 10 Important Features (Averaged Across 24 Hours):")
        for idx, row in feature_importance_df.head(10).iterrows():
            print(f"  {row['Feature']}: {row['Mean_Abs_SHAP']:.4f}")
        
        shap_result = {
            'shap_values_averaged': averaged_shap_values,
            'expected_value_averaged': averaged_expected_value,
            'all_shap_values': all_shap_values,  # Keep individual hours for waterfall
            'all_expected_values': all_expected_values,
            'feature_importance': feature_importance_df,
            'X_sample': X_test_sample,
            'y_sample': y_test_sample,
            'y_pred_sample': y_pred_sample
        }
        
        # Generate SHAP plots
        self._plot_shap_summary(shap_result)
        self._plot_shap_dependence(shap_result)
        self._plot_shap_waterfall(shap_result)
        
        return shap_result
    
    def _plot_shap_summary(self, shap_result: Dict):
        """SHAP Summary Plot (Averaged, Top 20 features)"""
        print("\n" + "=" * 100)
        print("【SHAP Summary Plot (Averaged Across 24 Hours)】")
        print("=" * 100)
        
        fig = plt.figure(figsize=(14, 10))
        
        shap.summary_plot(
            shap_result['shap_values_averaged'],
            shap_result['X_sample'],
            plot_type="dot",
            show=False,
            max_display=20
        )
        
        plt.title(f'SHAP Summary Plot - {self.target_site} (Averaged Across All 24 Hours)', 
                 fontweight='bold', fontsize=15, pad=20)
        plt.xlabel('SHAP Value (Impact on Model Output)', fontweight='bold', fontsize=13)
        plt.ylabel('Features', fontweight='bold', fontsize=13)
        
        plt.tight_layout()
        filename = f'{self.target_site}_SHAP_Summary_Plot.png'
        plt.savefig(filename, dpi=300, bbox_inches='tight')
        print(f"✓ Summary plot saved: {filename}")
        plt.show()
    
    def _plot_shap_dependence(self, shap_result: Dict):
        """SHAP Dependence Plots (Top 4 features from averaged SHAP)"""
        print("\n" + "=" * 100)
        print("【SHAP Dependence Plots】")
        print("=" * 100)
        
        top_4_features = shap_result['feature_importance'].head(4)['Feature'].tolist()
        
        print(f"➤ Top 4 Features (from averaged SHAP):")
        for i, feat in enumerate(top_4_features, 1):
            print(f"  {i}. {feat}")
        
        fig, axes = plt.subplots(2, 2, figsize=(18, 12))
        axes = axes.flatten()
        
        for idx, feature in enumerate(top_4_features):
            ax = axes[idx]
            
            shap.dependence_plot(
                feature,
                shap_result['shap_values_averaged'],
                shap_result['X_sample'],
                show=False,
                ax=ax
            )
            
            ax.set_title(f'({chr(65+idx)}) SHAP Dependence - {feature}', 
                        fontweight='bold', fontsize=13, pad=10, loc='left')
            ax.set_xlabel(f'{feature} (Feature Value)', fontweight='bold', fontsize=11)
            ax.set_ylabel('SHAP Value (Averaged)', fontweight='bold', fontsize=11)
            ax.grid(True, alpha=0.3, linestyle='--')
        
        plt.suptitle(f'SHAP Dependence Plots - {self.target_site} (Averaged SHAP)', 
                    fontsize=16, fontweight='bold', y=0.995)
        plt.tight_layout()
        
        filename = f'{self.target_site}_SHAP_Dependence_Plots.png'
        plt.savefig(filename, dpi=300, bbox_inches='tight')
        print(f"✓ Dependence plots saved: {filename}")
        plt.show()
    
    def _plot_shap_waterfall(self, shap_result: Dict):
        """SHAP Waterfall Plots (Use Hour 1 for individual examples, Top 10 features)"""
        print("\n" + "=" * 100)
        print("【SHAP Waterfall Plots - Individual Predictions (Hour 1)】")
        print("=" * 100)
        
        # Use Hour 1 for waterfall plots
        hour_idx = self.config.shap_waterfall_hour
        hour_label = hour_idx + 1
        
        shap_values_hour1 = shap_result['all_shap_values'][hour_idx]
        expected_value_hour1 = shap_result['all_expected_values'][hour_idx]
        
        y_pred_sample = shap_result['y_pred_sample'][:, hour_idx]
        y_true_sample = shap_result['y_sample'][:, hour_idx]
        
        # Select representative samples
        high_aqi_idx = np.argmax(y_pred_sample)
        low_aqi_idx = np.argmin(y_pred_sample)
        mid_aqi_idx = np.argmin(np.abs(y_pred_sample - np.median(y_pred_sample)))
        
        cases = [
            {
                'index': high_aqi_idx,
                'label': 'High AQI Prediction',
                'filename': f'{self.target_site}_SHAP_Waterfall_High_AQI.png'
            },
            {
                'index': low_aqi_idx,
                'label': 'Low AQI Prediction',
                'filename': f'{self.target_site}_SHAP_Waterfall_Low_AQI.png'
            },
            {
                'index': mid_aqi_idx,
                'label': 'Medium AQI Prediction',
                'filename': f'{self.target_site}_SHAP_Waterfall_Medium_AQI.png'
            }
        ]
        
        print(f"\nSelected Representative Samples (Hour {hour_label}):")
        for case in cases:
            idx = case['index']
            pred = y_pred_sample[idx]
            actual = y_true_sample[idx]
            error = pred - actual
            
            print(f"  • {case['label']}")
            print(f"    Predicted: {pred:.2f}")
            print(f"    Actual: {actual:.2f}")
            print(f"    Error: {error:+.2f}\n")
        
        # Generate waterfall plot for each case
        for case in cases:
            sample_idx = case['index']
            
            # Create SHAP Explanation object
            explanation = shap.Explanation(
                values=shap_values_hour1[sample_idx],
                base_values=expected_value_hour1,
                data=shap_result['X_sample'].iloc[sample_idx].values,
                feature_names=shap_result['X_sample'].columns.tolist()
            )
            
            fig = plt.figure(figsize=(12, 8))
            
            # ========== Only show Top 10 features ==========
            shap.waterfall_plot(explanation, max_display=10, show=False)
            
            ax = plt.gca()
            
            pred = y_pred_sample[sample_idx]
            actual = y_true_sample[sample_idx]
            
            # English title
            title_text = f'{case["label"]} - {self.target_site} (Hour {hour_label})\n'
            title_text += f'Predicted: {pred:.2f} | Actual: {actual:.2f} | Error: {pred-actual:+.2f}\n'
            title_text += f'Baseline: {expected_value_hour1:.2f}'
            
            ax.set_title(title_text, fontweight='bold', fontsize=13, pad=20, loc='center')
            ax.set_xlabel('SHAP Value (Impact on Model Output)', fontweight='bold', fontsize=12, labelpad=10)
            ax.tick_params(axis='y', labelsize=10, pad=8)
            ax.grid(True, alpha=0.3, linestyle='--', axis='x')
            
            plt.tight_layout(pad=2.0)
            plt.savefig(case['filename'], dpi=300, bbox_inches='tight')
            print(f"  ✓ Saved: {case['filename']}")
            plt.show()
    
    # ==================== Main Run Function ====================
    
    def run(self) -> Dict:
        print(f"\n{'='*70}")
        print(f"🎯 Starting Prediction: {self.target_site}")
        print(f"{'='*70}")
        
        lag_config = self.analyze_correlations()
        df_features = self.prepare_features(lag_config)
        
        X, y = self.create_sequences(df_features)
        
        (X_train, y_train, X_val, y_val, X_test, y_test,
         y_train_raw, y_val_raw, y_test_raw) = self.prepare_data(X, y)
        
        print(f"\n📊 Data Split:")
        total = len(X_train) + len(X_val) + len(X_test)
        print(f"   Train: {len(X_train)} ({len(X_train)/total*100:.1f}%)")
        print(f"   Val: {len(X_val)} ({len(X_val)/total*100:.1f}%)")
        print(f"   Test: {len(X_test)} ({len(X_test)/total*100:.1f}%)")
        
        if self.config.use_optuna:
            self.optimizer = FastOptunaOptimizer(X_train, y_train, X_val, y_val, self.config)
            self.best_params = self.optimizer.optimize()
            lgb_params, n_estimators = self.optimizer.get_best_lgb_params()
        else:
            lgb_params = self.config.lgb_params
            n_estimators = self.config.num_boost_round
        
        self.train_final_model(X_train, y_train, X_val, y_val, lgb_params, n_estimators)
        
        y_pred_train = self.predict(X_train)
        y_pred_val = self.predict(X_val)
        y_pred_test = self.predict(X_test)
        
        train_eval = self.evaluate(y_train_raw, y_pred_train)
        val_eval = self.evaluate(y_val_raw, y_pred_val)
        test_eval = self.evaluate(y_test_raw, y_pred_test)
        
        # Print results
        print(f"\n{'='*70}")
        print(f"📊 Final Results: {self.target_site}")
        print(f"{'='*70}")
        print(f"{'Metric':<15} {'Train':<15} {'Val':<15} {'Test':<15}")
        print("-" * 60)
        print(f"{'RMSE':<15} {train_eval['overall']['rmse']:<15.4f} {val_eval['overall']['rmse']:<15.4f} {test_eval['overall']['rmse']:<15.4f}")
        print(f"{'MAE':<15} {train_eval['overall']['mae']:<15.4f} {val_eval['overall']['mae']:<15.4f} {test_eval['overall']['mae']:<15.4f}")
        print(f"{'R² (weighted)':<15} {train_eval['overall']['r2_weighted']:<15.4f} {val_eval['overall']['r2_weighted']:<15.4f} {test_eval['overall']['r2_weighted']:<15.4f}")
        print(f"{'R² (flat)':<15} {train_eval['overall']['r2_flat']:<15.4f} {val_eval['overall']['r2_flat']:<15.4f} {test_eval['overall']['r2_flat']:<15.4f}")
        
        print("\n【Period R² - Test Set】")
        for period, metrics in test_eval['period'].items():
            print(f"  {period}: {metrics['r2']:.4f}")
        
        # Overfitting detection
        print(f"\n🔍 Overfitting Detection:")
        train_test_gap = train_eval['overall']['r2_flat'] - test_eval['overall']['r2_flat']
        val_test_gap = val_eval['overall']['r2_flat'] - test_eval['overall']['r2_flat']
        print(f"   Train-Test R² Gap: {train_test_gap:+.4f}")
        print(f"   Val-Test R² Gap: {val_test_gap:+.4f}")
        
        if val_test_gap > 0.05:
            print("   ⚠️ Significant overfitting detected")
        elif val_test_gap > 0.02:
            print("   ⚡ Slight overfitting (acceptable)")
        else:
            print("   ✅ Good generalization")
        
        print(f"{'='*70}")
        
        # Detailed horizon analysis
        step_rmse, step_mae, horizon_results = self.detailed_horizon_analysis(
            y_test_raw, y_pred_test
        )
        
        # Plot prediction results
        self.plot_detailed_results(y_test_raw, y_pred_test, step_rmse, step_mae)
        
        # ========== SHAP Analysis ==========
        shap_result = self.analyze_shap(y_test_raw, y_pred_test)
        
        return {
            'test_metrics': test_eval,
            'train_metrics': train_eval,
            'val_metrics': val_eval,
            'best_params': self.best_params,
            'hourly_metrics': test_eval['hourly'],
            'step_rmse': step_rmse,
            'step_mae': step_mae,
            'horizon_results': horizon_results,
            'shap_result': shap_result
        }
    
    def save_model(self, path: str):
        joblib.dump({
            'model': self.model,
            'scaler_X': self.scaler_X,
            'scaler_y': self.scaler_y,
            'best_params': self.best_params,
            'config': self.config
        }, path)
        logger.info(f"Model saved: {path}")


def main():
    """Main Function"""
    DATA_PATH = 'C:/Users/32052/Desktop/小组/kaohsiung_5sites_final.csv'
    TARGET_SITE = 'Zuoying'
    
    config = ModelConfig(
        lookback_hours=48,
        forecast_hours=24,
        max_lag=24,
        test_size=0.15,
        val_size=0.15,
        num_boost_round=800,
        use_gpu=USE_GPU,
        use_optuna=True,
        n_trials=10,
        optuna_timeout=3600,
        optuna_sample_ratio=0.5,
        optuna_eval_hours=[0, 5, 11, 23],
        # SHAP Configuration
        use_shap=True,
        shap_sample_size=500,
        shap_waterfall_hour=0,  # Use Hour 1 for waterfall plots
    )
    
    predictor = FastPredictor(
        data_path=DATA_PATH,
        target_site=TARGET_SITE,
        config=config
    )
    
    result = predictor.run()
    
    # Save results
    result['hourly_metrics'].to_csv(f'{TARGET_SITE}_hourly_metrics.csv', index=False)
    
    horizon_df = pd.DataFrame(result['horizon_results']).T
    horizon_df.to_csv(f'{TARGET_SITE}_horizon_analysis.csv')
    
    step_df = pd.DataFrame({
        'hour': range(1, 25),
        'rmse': result['step_rmse'],
        'mae': result['step_mae']
    })
    step_df.to_csv(f'{TARGET_SITE}_hourly_errors.csv', index=False)
    
    if result['best_params']:
        pd.DataFrame([result['best_params']]).to_csv(f'{TARGET_SITE}_best_params.csv', index=False)
    
    # Save SHAP feature importance (averaged)
    if result['shap_result']:
        result['shap_result']['feature_importance'].to_csv(
            f'{TARGET_SITE}_SHAP_importance_averaged.csv',
            index=False
        )
    
    predictor.save_model(f'{TARGET_SITE}_model.pkl')
    
    print("\n✅ All Completed!")
    print(f"   Model saved: {TARGET_SITE}_model.pkl")
    print(f"   Results saved: {TARGET_SITE}_hourly_metrics.csv")
    print(f"   SHAP plots saved (averaged across 24 hours)")


if __name__ == "__main__":
    main()
```
